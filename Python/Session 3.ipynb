#### 4. Create a prime number function
>A prime number is greater than 1 and divisible only by 1 and itself. <br>
>The loop checks divisibility from 2 up to the square root of n (which is more efficient than checking all the way to n-1).

```python
def is_prime(n):
    if n <= 1:
        return  print(n, "is not a prime number")
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return  print(n, "is not a prime number")
    return print(n, "is a prime number")

# Example usage
is_prime(7)

```

```python
is_prime(25)
```

#### 5. Create a Fibonacci series function
The Fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones, typically starting with 0 and 1. The sequence goes 0, 1, 1, 2, 3, 5, 8, 13, and so on, 

```python
def fibonacci(n):
    a=0
    b=1
    for i in range(1,n+1):
        print(a)
        c=a+b
        a=b
        b=c
    

fibonacci(7)
```

#### while loop:A while loop in Python is used to repeatedly execute a block of code as long as a given condition is True.

```python
i =0
while i < 5:
    print('i is: {}'.format(i))
    i = i+1
```

```python
def fibonacci(n):
    a, b = 0, 1
    count = 0
    while count < n:
        print(a, end=" ")
        a, b = b, a + b
        count += 1

# Example usage
fibonacci(7)
```

### MAP function:
The map function applies a given function to all items in an input list and return a map object.<br>

map(function,inputlist)

```python
num=[1,2,3]
for i in num:
    print(square(i))
```

```python
x=[1,2,3]
result=[num**2 for num in x]
result
```

```python
num=[1,2,3]
a=list(map(square,num))
a
```

```python
num=[1,2,3]
a=list(map(lambda x:x*x,num))
a
```
