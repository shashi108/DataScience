# Python OOPs Concepts — Easy Explanation with Examples

OOPs stands for **Object-Oriented Programming**. It is a programming approach in Python where we organize code using **classes and objects**.

For example, imagine a car. Every car has properties such as color and brand, and behaviors such as starting and stopping. We can represent these using a Python class.

## 1. The 4 Main Pillars of OOPs

### 1. Encapsulation

Wrapping data and methods together inside a class.

**Example:** A bank account manages its balance through methods.

### 2. Inheritance

One class can reuse properties and methods from another class.

**Example:** A `Car` class inherits from a `Vehicle` class.

### 3. Polymorphism

The same method name can behave differently for different objects.

**Example:** Different animals implement the `sound()` method differently.

### 4. Abstraction

Hiding implementation details and exposing only the essential functionality.

**Example:** You drive a car without needing to understand how its engine works internally.

Why does it give an error?

When we call `com.config()`, Python automatically passes the object `com` as the first argument.

```
computer.config(com)
```

We use `self` as the first parameter.

````python
class computer():
    def config(self):
        print('i5', '8gb')

com = computer()
com.config()
````

Why does it work?

- Python automatically passes the object `com`.
- `self` receives that object.
- Therefore, there is no error.

### 3.3. Using Instance Variables (`self.cpu` and `self.ram`)

- This is the correct way to store the values passed during object creation.
- Instance variables are variables that belong to an individual object. Each object has its own copy of instance variables, so changing the value in one object does not affect another object.

````python
class computer():
    def __init__(self, cpu, ram):
        self.cpu = cpu    #hear self is the object & we need to refer to the object
        self.ram = ram

    def config(self):
        print("Configuration:", self.cpu, self.ram)

com1 = computer('i5', 8)
com2 = computer('i7', 16)

com1.config()
com2.config()
````

- `cpu` and `ram` are parameters of the constructor.
- `self.cpu` and `self.ram` are instance variables.
- `self.cpu = cpu` stores the CPU value inside the current object.
- `self.ram = ram` stores the RAM value inside the current object.

#### 3.4 Update Instance Variables 

````python
class computer():
    def __init__(self):
        self.name = 'akash'
        self.age = 37

c1 = computer()
c2 = computer()

c1.name = 'nit'
c1.age = 23

print(c1.name, c1.age)
print(c1.age)
print(c2.name)

print(id(c1))
print(id(c2))
````

In this code, we create a class `computer` with a constructor `__init__()` that initializes two instance variables, `name` and `age`, with default values `'akash'` and `37`. We then create two objects, `c1` and `c2`, of the same class. Initially, both objects have the same values. However, when we update `c1.name = 'nit'` and `c1.age = 23`, only the values of `c1` change, while `c2` remains unchanged because both objects maintain their own instance variables. The `print()` statements display the updated and original values, and the `id()` function returns the unique identity of each object, showing that `c1` and `c2` are two separate objects.

### 4.1. Instance Variable

An Instance Variable is a variable whose value is different for each object.

- Defined using `self`.
- Each object maintains its own values.
- Usually initialized inside the constructor (`__init__`).

class Student():

    def __init__(self, name, marks): # Instance method
        self.name = name       # Instance Variable
        self.marks = marks     # Instance Variable

s1 = Student("Ram", 90)
s2 = Student("Shyam", 80)

print(s1.name)
print(s2.name)

### 4.2. Class Variable

A Class Variable is a variable that belongs to the class and is shared by all objects.

- Defined inside the class, outside methods.
- Can be accessed using the class name or object name.
- Shared across objects unless an object creates its own attribute with the same name.

````python
class Student:

    school = "NIT"  # Class Variable

    def __init__(self, name): # Instance method
        self.name = name      # Instance Variable

s1 = Student("Ram")
s2 = Student("Shyam")

print(s1.school)
print(s2.school)
print(Student.school)
````

### 4.3. Instance Method

An Instance Method is a method that works with instance variables.

- Uses `self` as the first parameter.
- Can access instance and class variables.
- No decorator is required.

````python
class student():  
        
    school = 'nit'   #school is class variable
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable (m1,m2,m3 are instance variable)
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
#lets create a method to calculate average of the marks 
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(s2.avg())
````

### 4.4. Class Method

A Class Method is a method that works with class-level data.

- Uses `cls` as the first parameter.
- Uses the `@classmethod` decorator.
- Can access and modify class variables.
- Can be called using the class name or object name.

````python
class Student:

    school = "NIT"  #school is class variable

    @classmethod
    def info(cls):
        return cls.school

print(Student.info())
````

### 4.5. What is a Static Method?

A Static Method is a method inside a class that does not depend on instance variables or class variables.

- It does not access any variables-class or instance 
- We use the `@staticmethod` decorator.
- It does not require `self` or `cls`.
- It cannot directly access instance or class variables.
- It can be called using the class name or object name.
- It is used for utility or general-purpose operations related to a class.


````python
class Student:

    @staticmethod
    def welcome():
        print("Welcome to Python OOP")

# Calling Static Method
Student.welcome()
````

### Inner Class (Nested Class)
An **Inner Class** is a class defined inside another class. In this code, `student` is the outer class, and `LAPTOP` is the inner class. The student class contains the student's name and roll number, while the laptop class contains the laptop's brand, CPU, and RAM. Inside the student constructor, `self.lap = self.LAPTOP()` creates a separate laptop object for every student object. Therefore, when we create `s1` and `s2`, each student gets their own laptop object. We can access the laptop's properties using `s1.lap.cpu`, `s2.lap.ram`, or `s2.lap.brand`. We can also create a laptop object directly using `student.LAPTOP()`.

````python
class student:
    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno
        
        self.lap = self.LAPTOP()
        
    def show(self):
        print(self.name, self.rollno)
        
    class LAPTOP:
        def __init__(self):
            self.brand = 'avita'
            self.cpu = 'i5'
            self.ram = 8
     
s1 = student('a',1) # the moment when we assign the variable we need to assign the class as well 
s2 = student('b',2)

s1.show()
s2.show()

lap1 = s1.lap.cpu
lap2 = s2.lap.ram
lap3 = s2.lap.brand

print(lap1)
print(lap2)
print(lap3)
````

### 5.What is a Decorator?

A **Decorator** is something that modifies or extends the behaviour of a function or method without changing its original code.

In simple words, a decorator tells Python how a particular function should behave.

- We use the `@` symbol to apply a decorator.
- It helps extend existing functionality.
- It avoids modifying the original function.
- It promotes code reusability.

````python
class student:
    school = 'nit'
    def info(self):
        return self.school
````

Here, info() is an instance method because it takes self as its first parameter.

**With decorator:**

````python
class student:
    school = 'nit'
    @classmethod
    def info(cls):
        return cls.school
````

@classmethod is a built-in Python decorator that converts a method into a class method.Here, @classmethod changes the method's behaviour.
- self refers to an object.
- cls refers to a class.

- The @my_decorator syntax automatically passes the welcome function to the decorator.

````python
@my_decorator
def welcome():
    print("Welcome")

#The above code is equivalent to:

def welcome():
    print("Welcome")
welcome = my_decorator(welcome)
````

````python
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper

@my_decorator
def welcome():
    print("Welcome to Python")


welcome()
````

### 6.1.Single level Inheritance

````python
class A(): 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

     # the way you access the feature is object 

a1 = A()  #a1 is the object & A is constructor & using a1 we can call the feature1 & feature 2

a1.feature1()
a1.feature2()
````

#### Visual Understanding 

```text
             Class A
           (Parent Class)
                 |
          --------------
          |            |
      feature1()    feature2()
                 |
                 | Inheritance
                 ↓
             Class B
           (Child Class)
                 |
       ---------------------
       |                   |
   Inherited Methods    Own Methods
       |                   |
   feature1()           feature3()
   feature2()           feature4()
```

````python
class A(): # parent class 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

class B(A): #child class
    def feature3(self):         
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')

a1 = A()  
b1 = B()

a1.feature1()
a1.feature2()
#---- Class B access the method of class A also
print("---------")
b1.feature1()
b1.feature2()

b1.feature3()
b1.feature4()
````

### 6.2 Multilevel Inheritance

              Class A
           (Grandparent)
                  |
          ----------------
          |              |
      feature1()      feature2()
                  |
                  | Inheritance
                  ↓
              Class B
             (Parent Class)
                  |
          ----------------
          |              |
      feature3()      feature4()
                  |
                  | Inheritance

````python
class A(): # Grand parent class 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

class B(A): #Parent class
    def feature3(self):         
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')

class C(B): #Child class
    def feature5(self):
        print('bye')

a1 = A()  
b1 = B()
c1 = C()

print("---- Class A method ----")
a1.feature1()
a1.feature2()

print("---- Class B access the method of class A also ----")
b1.feature1()
b1.feature2()
b1.feature3()
b1.feature4()

print("---- Class C access the method of class A+B also ----")
c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
````

### 6.3 Multiple Inheritance

````python
       Class A                 Class B
      (Parent 1)              (Parent 2)
          |                       |
     -----------              -----------
     |         |              |         |
 feature1() feature2()    feature3() feature4()
          \                       /
           \                     /
            \                   /
                  Class C
                  (Child)
                     |
                  feature5()
````

````python
class A(): # parent class 
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

class B(): #child class
    def feature3(self):         
        print('feature3 is hello')
    def feature4(self):
        print('feature4 is hello')

class C(A,B): #MULTIPLE INHERITENCE
    def feature5(self):
        print('bye')

a1 = A()  
b1 = B()
c1 = C()

print("---- Class A method ----")
a1.feature1()
a1.feature2()

print("---- Class B method  ----")

b1.feature3()
b1.feature4()

print("---- Class C access the method of class A+B also ----")
c1.feature1()
c1.feature2()
c1.feature3()
c1.feature4()
c1.feature5()
````

### 6.4 Hybrid Inheritance

Hybrid Inheritance is a combination of two or more types of inheritance, such as Hierarchical + Multiple Inheritance.

````python
class A:  # Parent class
    def feature1(self):
        print("Feature 1 from Class A")


class B(A):  # Hierarchical / Single inheritance
    def feature2(self):
        print("Feature 2 from Class B")


class C(A):  # Hierarchical inheritance
    def feature3(self):
        print("Feature 3 from Class C")


class D(B, C):  # Multiple inheritance
    def feature4(self):
        print("Feature 4 from Class D")


obj = D()

obj.feature1()
obj.feature2()
obj.feature3()
obj.feature4()
````

#### 6.6 Super() in Inhertience

`super().__init__()` is used to call the parent class constructor from the child class.

It executes the parent constructor first, then continues executing the child constructor.

````python
# constructor in inheritence 

class A() :  #super class
    def __init__(self):
        print('in A init')
    def feature1(self):         
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

class B(A):     #sub class
    def __init__(self):
        super().__init__()
        print('in B init')
    def feature3(self):         
        print('feature3 is hallo')
    def feature4(self):
        print('feature4 is hallo')
        
b1 = B() 
````

#### 6.7 MRO (Method Resolution Order)

MRO (Method Resolution Order) defines the order in which Python searches for methods and constructors in multiple inheritance.

In `class C(A, B)`, the search order is C → A → B → object. `super()` follows this order.

````python
# MRO (method resolution order) 

class A() : 
    def __init__(self):
        print('in the A init')
    def feature1(self):         # lets take example class A has 2 feature1 & feature 2
        print('feature1 is hi')
    def feature2(self):
        print('feature2 is hi')

class B():
    def __init__(self):
        print('in the B init')
    def feature3(self):         # lets take example class A has 2 feature1 & feature 2
        print('feature3 is hallo')
    def feature4(self):
        print('feature4 is hallo')
        
class C(A,B): 
    def __init__(self):
        super().__init__()
        print('in the C init')
                
c1 = C()
b1 = B()
````

## 7.1. Duck Typing

Duck typing means Python focuses on what an object can do rather than its class or type.

````python
class Dog:
    def speak(self):
        return "Woof!"

class Cat:
    def speak(self):
        return "Meow!"

class Duck:
    def speak(self):
        return "Quack!"

def animal_sound(animal):
    return animal.speak()

# Using duck typing
dog = Dog()
cat = Cat()
duck = Duck()

print(animal_sound(dog))  # Output: Woof!
print(animal_sound(cat))  # Output: Meow!
print(animal_sound(duck)) # Output: Quack!
````

Both objects have the same method `speak()`, but they produce different outputs.

## 7.3. Method Overloading

Method overloading means defining multiple methods with the same name but different parameters.

Python does not directly support traditional method overloading like Java or C++. However, we can achieve similar behaviour using default arguments or variable-length arguments.


````python
class demo():
    def add(self,a,b):
        return a+b

d = demo()
print(d.add(2,3))
````

````python
# this concept called as method overloading (same method different argument)
# python doesnot support python overloading we can define same but it will take latest method
# discard the first method which contrain 2 argument
class demo:
    def add(self,a,b):
        return a+b

    def add(self,a,b,c):
        return a+b+c

d = demo()

print(d.add(2,3))
print(d.add(4,5,6))
````

````python
# this method overloading we can acheiving with different concept using default argument

class demo:
    def add(self,a,b,c=0, d=0):
        return a+b+c+d

d = demo()

print(d.add(2,3))
print(d.add(4,5,6))
print(d.add(4,5,6,7))
````

The same method `add()` works with different numbers of arguments is called as method overloading

````python
class demo:
    def add(self,*arg):
        total = 0
        for i in arg:
            total = total + i
        return total

d = demo()

print(d.add(2,3))
print(d.add(4,5,6))
print(d.add(1,2,3,4,5,6))
````

## 7.4. Method Overriding

Method overriding occurs when a child class defines a method that already exists in its parent class.

The child class provides its own implementation of that method.

````python
class Parent:
    def show(self):
        print("This is Parent class")

class Child(Parent):
    def show(self):
        print("This is Child class")

obj = Child()
obj.show()
````

Here, the child class overrides the `show()` method of the parent class.

## 8.Abstract Method and Abstract Class

**8.1. Abstract Method:**

An **abstract method** is a method that is declared in the parent class but does not have any implementation. The child class must provide the implementation of that method.

**8.2. Abstract Class:**

A class that contains at least one abstract method is called an **abstract class**. We cannot create an object of an abstract class directly.

````python
class computer():
    def process(self):
        pass 
    
com = computer()
com.process()
````

- in this above code we have only method 'process' & we dont have body inside the method
- this method only has declaration but not defination we calld them as ABSTRACT METHOD

````python
from abc import ABC, abstractmethod

class computer(ABC):
    @abstractmethod  # AS DECORATOR   
    def process(self):
        pass
    
com = computer()
com.process()

#missing child class with same method
````

````python
from abc import ABC, abstractmethod # NEED TO IMPORT ABSTRACT CLASS & ABSTRACT METHOD

class computer(ABC): # --> ABSTRACT CLASS
    @abstractmethod # --> ABSTRACT METHOD USING DECORATOR
    def process(self):
        pass

class Laptop(computer):    #------ Laptop is subclass of computer 
    pass                   # same method but missing defination
    
com1 = Laptop()
com = computer()
com.process()
````

````python
from abc import ABC, abstractmethod

class computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(computer):    #------ Laptop is subclass of computer 
    def process(self):     # same method but having defination
        print('its working')
        
com1 = Laptop()
com1.process()
````

````python
from abc import ABC, abstractmethod

class Animal(ABC):  # Abstract class
    @abstractmethod
    def sound(self):  # Abstract method
        pass

class Dog(Animal):
    def sound(self):  # same method but having defination
        print("Bark")

class Cat(Animal):
    def sound(self):  # same method but having defination
        print("Meow")

d = Dog()
d.sound()

c = Cat()
c.sound()
````

1. `Animal` is an **abstract class** because it contains an abstract method `sound()`.
2. `sound()` is an **abstract method** because it has no implementation in the parent class.
3. `Dog` and `Cat` are child classes of `Animal`.
4. Both child classes provide their own implementation of the `sound()` method.
5. We cannot create an object of `Animal` directly.

````python
class topten():
    
    def __init__(self):
        self.num = 1
        
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.num <= 10:
            val = self.num
            self.num += 1
            return val
        else:
            raise StopIteration      # only way to stop the for loop is raise the exception          
    
values = topten()

#print(next(values))

for i in values:
    print(i)  
````

## 10.Generator in Python

A **generator** is a special type of function that returns values one by one instead of returning all values at once.
In Python, we use the `yield` keyword to create a generator.

````python
def topten():
    return 5
   
values = topten()

print(values) 
````

````python
def topten():
    yield 5
    
values = topten()

print(values)
````

````python
print(next(values))
````

````python
# Lets print top 10 perfect square
def topten():
    n = 1
    while n<=10:
        sq = n*n
        yield sq
        n += 1
    
values = topten()
    
for i in values:
    print(i)
````

## 11. Multithreading Concepts in Python

### 11.1. What is Multithreading?

**Multithreading** is a concept in Python that allows multiple threads to execute concurrently within a single program.

A **thread** is the smallest unit of execution within a process.

Instead of executing one task at a time, multithreading allows a program to handle multiple tasks concurrently.

**Example:** Imagine downloading three files. Without multithreading, Python downloads them one by one. With multithreading, it can manage all three downloads concurrently.

### 11.2. What is a Thread?

A thread is an independent sequence of execution within a process.

A Python program starts with one main thread. We can create additional threads to perform different tasks.

### 11.3. Process vs Thread

| Process | Thread |
|---|---|
| An independent program in execution. | A unit of execution within a process. |
| Has its own memory space. | Shares memory with other threads in the process. |
| More resource-intensive. | Generally lightweight. |
| Communication is comparatively difficult. | Communication is easier. |
