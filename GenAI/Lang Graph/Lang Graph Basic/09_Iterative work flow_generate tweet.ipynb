# Generate Tweet Using Iterative Workflow

## 🎯 Objective

Create a **tweet generation workflow** where the tweet is:

1. Generated
2. Evaluated
3. Improved if required
4. Evaluated again
5. Repeated until the tweet is approved

---

## 🔄 Iterative Workflow

```text
START
  ↓
Generate Tweet
  ↓
Evaluate Tweet
  ↓
 ┌──────────────────────┐
 │   Is Tweet Approved? │
 └──────────────────────┘
       ↓           ↓
     YES           NO
      ↓             ↓
     END       Optimize Tweet
                    ↓
                Evaluate
                    ↓
              (Repeat...)
```


````python
#Define Library
from langgraph.graph import StateGraph, START, END
from typing import TypedDict,Literal,Annotated
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage
import operator
````

````python
#local ollama
from langchain_ollama import ChatOllama
model = ChatOllama(model="gemma3:4b")
response = model.invoke("the capital of India? in one word")
print(response.content)
````

````python
generator_llm = ChatOllama(model="qwen3:4b")
evaluator_llm = ChatOllama(model="qwen3:8b")
optimizer_llm = ChatOllama(model="qwen3:8b")
````

````python
evaluator_llm .invoke("the capital of India? in one word").content
````

````python
class TweetState(TypedDict):
    topic:str
    tweet:str
    evaluation:Literal["approved","needs_improvement"]
    feedback:str
    iteration:int              # show number of looping
    max_iteration:int          #avoid infinite looping

    tweet_history: Annotated[list[str], operator.add]
    feedback_history: Annotated[list[str], operator.add]
````

````python
def generate_tweet(state: TweetState):

    # prompt
    messages = [
        SystemMessage(content="You are a funny and clever Twitter/X influencer."),
        HumanMessage(content=f"""
Write a short, original, and hilarious tweet on the topic: "{state['topic']}".

Rules:
- Do NOT use question-answer format.
- Max 280 characters.
- Use observational humor, irony, sarcasm, or cultural references.
- Think in meme logic, punchlines, or relatable takes.
- Use simple, day to day english
""")
    ]

    # send generator_llm
    response = generator_llm.invoke(messages).content

    # return response
    return {'tweet': response, 'tweet_history': [response]}
````

````python
from pydantic import BaseModel, Field

class TweetEvaluation(BaseModel):
    evaluation: Literal["approved", "needs_improvement"] = Field(..., description="Final evaluation result.")
    feedback: str = Field(..., description="feedback for the tweet.")
structured_evaluator_llm = evaluator_llm.with_structured_output(TweetEvaluation)
````

````python
def evaluate_tweet(state: TweetState):

    # prompt
    messages = [
    SystemMessage(content="You are a ruthless, no-laugh-given Twitter critic. You evaluate tweets based on humor, originality, virality, and tweet format."),
    HumanMessage(content=f"""
Evaluate the following tweet:

Tweet: "{state['tweet']}"

Use the criteria below to evaluate the tweet:

1. Originality – Is this fresh, or have you seen it a hundred times before?  
2. Humor – Did it genuinely make you smile, laugh, or chuckle?  
3. Punchiness – Is it short, sharp, and scroll-stopping?  
4. Virality Potential – Would people retweet or share it?  
5. Format – Is it a well-formed tweet (not a setup-punchline joke, not a Q&A joke, and under 280 characters)?

Auto-reject if:
- It's written in question-answer format (e.g., "Why did..." or "What happens when...")
- It exceeds 280 characters
- It reads like a traditional setup-punchline joke
- Dont end with generic, throwaway, or deflating lines that weaken the humor (e.g., “Masterpieces of the auntie-uncle universe” or vague summaries)

### Respond ONLY in structured format:
- evaluation: "approved" or "needs_improvement"  
- feedback: One paragraph explaining the strengths and weaknesses 
""")
]

    response = structured_evaluator_llm.invoke(messages)

    return {'evaluation':response.evaluation, 'feedback': response.feedback, 'feedback_history': [response.feedback]}
````

````python
def optimize_tweet(state: TweetState):

    messages = [
        SystemMessage(content="You punch up tweets for virality and humor based on given feedback."),
        HumanMessage(content=f"""
Improve the tweet based on this feedback:
"{state['feedback']}"

Topic: "{state['topic']}"
Original Tweet:
{state['tweet']}

Re-write it as a short, viral-worthy tweet. Avoid Q&A style and stay under 280 characters.
""")
    ]

    response = optimizer_llm.invoke(messages).content
    iteration = state['iteration'] + 1

    return {'tweet': response, 'iteration': iteration, 'tweet_history': [response]}
````

````python
def route_evaluation(state: TweetState):

    if state['evaluation'] == 'approved' or state['iteration'] >= state['max_iteration']:
        return 'approved'
    else:
        return 'needs_improvement'
````

````python
graph=StateGraph(TweetState)

#node
graph.add_node('generate',generate_tweet)
graph.add_node('evaluate',evaluate_tweet)
graph.add_node('optimize',optimize_tweet)

#edge
graph.add_edge(START,'generate')
graph.add_edge('generate','evaluate')

#Syntax: graph.add_conditional_edges("current_node",routing_function,{"condition_1": "next_node_1","condition_2": "next_node_2"})
graph.add_conditional_edges('evaluate', route_evaluation, {'approved': END, 'needs_improvement': 'optimize'})
graph.add_edge('optimize', 'evaluate')

workflow = graph.compile()

workflow
````

````python
initial_state = {
    "topic": "Lang Chain",
    "iteration": 1,
    "max_iteration": 5
}
result = workflow.invoke(initial_state)
result
````

# `add_conditional_edges()` – Case 1 vs Case 2

In LangGraph, `add_conditional_edges()` is used when we want to **choose the next node based on a condition**.

---

## Case 1 – Return value is the node name

```python
graph.add_conditional_edges(
    'calculate_discriminant',
    check_condition
)
```

Here, the `check_condition()` function returns the **actual node name**.

### Example

```python
def check_condition(state):
    D = state["discriminant"]

    if D > 0:
        return "real_roots"
    elif D == 0:
        return "repeated_roots"
    else:
        return "no_real_roots"
```

The returned values are already the node names:

```text
"real_roots"       → real_roots
"repeated_roots"   → repeated_roots
"no_real_roots"    → no_real_roots
```

Therefore, we don't need a mapping dictionary.

### Flow

```text
calculate_discriminant
          ↓
   check_condition()
     /      |       \
    ↓       ↓        ↓
real_roots repeated no_real_roots
```

---

# Case 2 – Return value is a condition/key

```python
graph.add_conditional_edges(
    'evaluate',
    route_evaluation,
    {
        'approved': END,
        'needs_improvement': 'optimize'
    }
)
```

Here, `route_evaluation()` returns a **condition/key**.

### Example

```python
def route_evaluation(state):

    if state["approved"]:
        return "approved"
    else:
        return "needs_improvement"
```

The mapping dictionary tells LangGraph where to go:

```python
{
    "approved": END,
    "needs_improvement": "optimize"
}
```

So:

```text
"approved"
     ↓
    END
```

and

```text
"needs_improvement"
     ↓
  optimize
```

### Flow

```text
       evaluate
           ↓
   route_evaluation()
       /        \
      /          \
 approved    needs_improvement
    ↓                ↓
   END            optimize
                    ↓
                 evaluate
```

---

# Why not always use Case 2?

We **can** use Case 2 for Case 1.

For example:

```python
graph.add_conditional_edges(
    'calculate_discriminant',
    check_condition,
    {
        'real_roots': 'real_roots',
        'repeated_roots': 'repeated_roots',
        'no_real_roots': 'no_real_roots'
    }
)
```

This is completely valid.

However, it is **redundant** because the condition names and node names are exactly the same.

```text
Condition          Node

real_roots     →   real_roots
repeated_roots →   repeated_roots
no_real_roots  →   no_real_roots
```

So we can simply write:

```python
graph.add_conditional_edges(
    'calculate_discriminant',
    check_condition
)
```

---

# Easy Rule to Remember

### Case 1

If the routing function returns the **node name**:

```python
return "real_roots"
```

Use:

```python
graph.add_conditional_edges("A", route_function)
```

---

### Case 2

If the routing function returns a **condition/key** that is different from the node name:

```python
return "approved"
```

Use a mapping:

```python
graph.add_conditional_edges(
    "A",
    route_function,
    {
        "approved": END,
        "needs_improvement": "optimize"
    }
)
```

---

## ⭐ In one line

> **Case 1:** Return value = destination node name → mapping is optional/unnecessary.

> **Case 2:** Return value = condition/key → mapping tells LangGraph the destination node.
